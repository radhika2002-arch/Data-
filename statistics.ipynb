```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
data=pd.read_csv('X.csv')
data.head()
```

```python
data['x'].mean()
```

```python
data['x'].median()
```

```python
data['x'].mode()[0]
```

```python
# data['x].sts()
variance=0
mean_value=54.41919999999999
count=50
for i in data['x'].values:
    variance+=(i-mean_value)**2/count
print(variance)
std=np.sqrt(variance)   
std 
```

```python
# Z_score
# (x-mean)/std
Zscore=[]
for i in data['x'].values:
    Zscore_value=(i-mean_value)/std 
    Zscore.append(Zscore_value)
Zscore 
```

```python
from scipy.stats import zscore
```

```python
Zscore_value=zscore(data['x'])
Zscore_value
```
